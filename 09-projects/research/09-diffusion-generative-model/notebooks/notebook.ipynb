# Denoising Diffusion Probabilistic Model (DDPM)
Forward noise schedule $q(x_t|x_0)$ and time-conditioned denoiser.


```python
import torch
from diffusion import GaussianDiffusion, MiniUNetDenoiser

diff = GaussianDiffusion(timesteps=200)
print('Alpha cumprod start:', diff.alphas_cumprod[0].item())
print('Alpha cumprod end  :', diff.alphas_cumprod[-1].item())
```

```python
x0 = torch.randn(4, 1, 16, 16)
t = torch.tensor([10, 50, 100, 199])
xt, noise = diff.q_sample(x0, t)

model = MiniUNetDenoiser(in_channels=1)
pred_noise = model(xt, t)
loss = torch.nn.functional.mse_loss(pred_noise, noise)
print(f'Initial Denoiser MSE Loss: {loss.item():.4f}')
```
