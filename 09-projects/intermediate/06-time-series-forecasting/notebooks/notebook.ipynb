# Time-Series Demand Forecasting
Autoregressive feature engineering, harmonic decomposition, and load forecasting.


```python
from forecaster import generate_load_series, TimeSeriesForecaster

series = generate_load_series(800)
print('Series summary:')
print(series.describe())
```

```python
forecaster = TimeSeriesForecaster()
forecaster.fit(series.iloc[:600])
metrics = forecaster.evaluate(series.iloc[600:])
print(f'Evaluation Metrics -> MAPE: {metrics["mape"]:.2f}%, RMSE: {metrics["rmse"]:.2f} kW')
```
