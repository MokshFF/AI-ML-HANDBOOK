# Production Image Classifier
End-to-end PyTorch residual convolutional network training and evaluation.


```python
import torch
from dataset import get_dataloaders
from classifier import ConvNet, train_and_eval

train_loader, test_loader = get_dataloaders(batch_size=32)
print('Train batches:', len(train_loader))
print('Test batches:', len(test_loader))
```

```python
model, acc = train_and_eval(epochs=5)
print(f'Test Top-1 Accuracy: {acc*100:.2f}%')
```

```python
imgs, labels = next(iter(test_loader))
with torch.no_grad():
    logits = model(imgs)
    preds = logits.argmax(dim=-1)
print('Sample Ground Truth:', labels[:8].tolist())
print('Sample Predictions :', preds[:8].tolist())
```
