# Two-Stage Recommendation System
Matrix factorization retrieval and ranking evaluation.


```python
from recommender import TwoStageRecommender, generate_mock_interactions

interactions = generate_mock_interactions(n_users=100, n_items=50, n_records=1000)
print('Total interactions:', len(interactions))
```

```python
rec = TwoStageRecommender(n_users=100, n_items=50, embed_dim=16)
rec.fit(interactions[:800], epochs=15)
print('Model fitted successfully.')
```

```python
recs = rec.recommend(user_id=12, top_k=5)
print('Top 5 recommendations for User 12:')
for rank, (item_id, score) in enumerate(recs, 1):
    print(f'{rank}. Item {item_id:02d} (Score: {score:.4f})')
```
