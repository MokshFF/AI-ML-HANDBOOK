# NLP Sentiment Classifier
Text tokenization, TF-IDF vectorization, and sentiment classification.


```python
from sentiment_data_loader import generate_sentiment_data
from sentiment_engine import SentimentPipeline

texts, labels = generate_sentiment_data(300)
print('Sample text:', texts[0], '-> Label:', labels[0])
```

```python
pipe = SentimentPipeline()
pipe.fit(texts[:240], labels[:240])
m = pipe.evaluate(texts[240:], labels[240:])
print('Test Metrics:', m)
```

```python
test_phrases = ['This is fantastic super fast shipping', 'Worst product ever completely broke']
preds, probs = pipe.predict(test_phrases)
for t, p, pr in zip(test_phrases, preds, probs):
    print(f'"{t}" -> {"Positive" if p==1 else "Negative"} (Confidence: {pr:.3f})')
```
