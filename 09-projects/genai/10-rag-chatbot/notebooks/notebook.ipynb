# Production RAG Chatbot
Chunking, vector embeddings, cosine retrieval, and grounded citation generation.


```python
from rag_pipeline import MockDenseEmbedder, InMemoryVectorStore, RAGChatbot

embedder = MockDenseEmbedder(dim=32)
store = InMemoryVectorStore(embedder)
store.add_documents([
    {'id': 'k8s', 'title': 'Kubernetes', 'content': 'Pods run containers and are grouped into deployments for horizontal scaling.'},
    {'id': 'db', 'title': 'PostgreSQL', 'content': 'Relational databases use ACID transactions and B-Tree indexes for fast point queries.'}
])
print('Vector store indexed docs:', len(store.docs))
```

```python
bot = RAGChatbot(store)
res = bot.answer_query('How do deployments scale pods?')
print('Answer:', res['answer'])
print('Citations:', res['citations'])
```
