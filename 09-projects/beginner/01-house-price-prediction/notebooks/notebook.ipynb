# End-to-End House Price Prediction
Interactive exploratory analysis and ridge regression modeling.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from housing_data_loader import generate_housing_data
from model import HousePricePredictor

df = generate_housing_data(1000)
df.head()
```

```python
print('Summary statistics:')
df.describe()
```

```python
train_df = df.iloc[:800]
test_df = df.iloc[800:]

predictor = HousePricePredictor(alpha=5.0)
predictor.fit(train_df)
metrics = predictor.evaluate(test_df)
print(f'Test Evaluation -> MAE: ${metrics["mae"]:,.2f}, RMSE: ${metrics["rmse"]:,.2f}, R2: {metrics["r2"]:.4f}')
```

```python
preds = predictor.predict(test_df)
residuals = test_df['sale_price'].values - preds
print(f'Mean residual: {residuals.mean():.2f}, Std residual: {residuals.std():.2f}')
```
