# Customer Churn Prediction
End-to-end binary classification with logistic regression and threshold analysis.


```python
import numpy as np
import pandas as pd
from churn_data_loader import generate_churn_data
from churn_model import CustomerChurnModel

df = generate_churn_data(1000)
print('Class distribution:')
print(df['churn'].value_counts(normalize=True))
```

```python
train_df = df.iloc[:800]
test_df = df.iloc[800:]

model = CustomerChurnModel(lr=0.1, n_epochs=300)
model.fit(train_df)
metrics = model.evaluate(test_df)
print('Default (0.50) Evaluation Metrics:', metrics)
```

```python
for thresh in [0.3, 0.4, 0.5, 0.6, 0.7]:
    m = model.evaluate(test_df, threshold=thresh)
    print(f'Threshold={thresh:.2f} -> Precision={m["precision"]:.3f}, Recall={m["recall"]:.3f}, F1={m["f1"]:.3f}')
```
