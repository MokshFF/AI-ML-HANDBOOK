# Text Preprocessing: Tokenization, Stemming, Bag-of-Words & TF-IDF

This interactive lab covers:
1. **Text Normalization & Tokenization**: Rule-based text cleaning and token extraction.
2. **Morphological Stemming**: Suffix stripping via Porter-style transformation rules.
3. **Bag of Words (CountVectorizer)**: Constructing document-term frequency matrices.
4. **TF-IDF & Semantic Document Similarity**: Weighting rare informative terms and computing cosine similarity.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd() / 'code'))

import numpy as np
from text_processor import SimpleTokenizer, PorterStemmerMini, BagOfWords, TfidfVectorizerScratch

print('Text Preprocessing modules loaded successfully!')
```

## 1. Tokenization & Morphological Stemming

We compare raw text against normalized tokens and apply suffix stripping to reduce inflected words to root stems.

```python
tokenizer = SimpleTokenizer(lowercase=True, remove_stopwords=True)
stemmer = PorterStemmerMini()

sample_text = 'The engineers were programming, running experiments, and designing distributed systems.'
tokens = tokenizer.tokenize(sample_text)
stemmed = [stemmer.stem(t) for t in tokens]

print(f'Raw Tokens (Stopwords Removed): {tokens}')
print(f'Stemmed Roots:                  {stemmed}')
```

## 2. Bag of Words (Term Frequencies)

We transform a corpus of documents into a document-term frequency matrix.

```python
corpus = [
    'machine learning systems require high quality training data',
    'deep neural networks learn hierarchical representations from data',
    'distributed systems enable scalable machine learning pipelines'
]

bow = BagOfWords()
count_matrix = bow.fit_transform(corpus)

print(f'Vocabulary Size: {len(bow.vocabulary_)}')
print('Document-Term Count Matrix Shape:', count_matrix.shape)
print('Sample Vocabulary Mapping:', dict(list(bow.vocabulary_.items())[:8]))
```

## 3. TF-IDF Representation & Information Retrieval

We compute smooth TF-IDF vectors and find the most relevant document for a query using cosine similarity.

```python
tfidf = TfidfVectorizerScratch(smooth_idf=True)
X_tfidf = tfidf.fit_transform(corpus)

query = 'scalable distributed pipelines'
q_vec = tfidf.transform([query])

# Pairwise Cosine Similarity: q_vec @ X_tfidf.T (since vectors are L2-normalized)
similarities = np.dot(q_vec, X_tfidf.T).flatten()
best_doc_idx = int(np.argmax(similarities))

print(f'Query: "{query}"')
for i, sim in enumerate(similarities):
    print(f'Doc {i} Similarity: {sim:.4f} -> "{corpus[i]}"')
print(f'\nTop Retrieved Document: Doc {best_doc_idx}')
assert best_doc_idx == 2, 'Information retrieval failed to select Doc 2!'
```
